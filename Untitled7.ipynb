```python
import pandas as pd

df = pd.DataFrame({
    'Age': [7,6,8],
    'Gender': ['M','F', 'M']
})
```

```python
print(df)
```

```python
cov_rx = df.cov()
```

```python
print(cov_rx)
```

```python
corr_matrix = df.corr()
print(corr_matrix)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
plt.title("Correlation Matrix Heatmap")
plt.show()
```
