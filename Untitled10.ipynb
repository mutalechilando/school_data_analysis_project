```python
import seaborn as sns

df = sns.load_dataset('titanic')
df.head()
```

```python
z = (df['age'] - df['age'].mean()) / df['age'].std() 
print(z)
```

```python
import matplotlib.pyplot as plt

sns.histplot(z, kde=True, color='teal')
plt.title('Distribution of Z-Scores for Age')
plt.xlabel('Z-score')
plt.ylabel('Frequency')
plt.show()
```

```python
outliers = df[abs(z) > 3]
print(outliers[['age']])
```
