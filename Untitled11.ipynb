```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('titanic')

# Step 1: Draw random samples and compute their means
means = [np.mean(np.random.choice(df['fare'].dropna(), 50)) for _ in range(1000)]
# Step 2: Plot the sampling distribution
sns.histplot(means, kde=True, color='skyblue')
plt.title("Sampling Distribution of Fare Means (n=50, 1000 samples)")
plt.xlabel("Sample Mean Fare")
plt.ylabel("Frequency")
plt.show()
```
