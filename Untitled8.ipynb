```python
import numpy as np

trials = 10
twos = 0

for i in range(trials):
    rolls = np.random.choice([1, 2, 3, 4, 5 , 6])
    if rolls == 2:
        twos += 1

prob_twos = twos / trials
print("Estimated P(Twos):", prob_twos)
```
