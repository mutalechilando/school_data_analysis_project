```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

students = pd.read_excel('students.xlsx')
df = pd.DataFrame(students)

print(df.head())



sns.histplot(df['Age'], bins=4, kde=True)
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.boxplot(x='Gender', y='Age', data=df)
plt.title("Boxplot of Age by Gender")
plt.show()
```

```python
df.head()
```

```python
sns.scatterplot(x='Age', y='Math_Score', hue='Class', data=df) 
plt.title("Scatter Plot of Age vs Math Score by Survival") 
plt.xlabel("Age") 
plt.ylabel("Math Score") 
plt.show() 
```

```python
sns.pairplot(df[['Age','Math_Score','Average_Score', 'Science_Score']]) 
plt.show() 
```
