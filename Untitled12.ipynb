```python
import pandas as pd
import numpy as np

np.random.seed(42)
temperature = np.random.randint(20, 40, 100)
ice_cream_sales = temperature * 10 + np.random.randint(0, 30, 100)
drowning_cases = temperature * 2 + np.random.randint(0, 5, 100)

df = pd.DataFrame({'temperature': temperature,
                   'ice_cream_sales': ice_cream_sales,
                   'drowning_cases': drowning_cases})

print(df.corr())


```
